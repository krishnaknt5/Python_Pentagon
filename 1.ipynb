```python
class Laptop:
    def __init__(self):
        self.color="Black"
        self.model="V15"
        self.brand="Asus"
    def run(self):
        print("Laptop is running")
        print(self)
l1=Laptop()
print(l1.color)
print(l1.model)
print(l1.brand)
l1.run()
print(l1)       
l1.proc="I5"
print(l1.proc)
```
