**No parameter with return value**

```python
class calci: 
    def __init__(self): 
        self.brand = "Casio" 
        self.color = "black" 
        
    def add(self): 
        a = 10 
        b = 20 
        c = a + b 
        return c 

C1 = calci() 
print(C1.brand) 
print(C1.color) 
res = C1.add()  
print(res)
```

**with parameter no return value**

```python
class calci: 
    def __init__(self): 
        self.brand = "Casio" 
        self.color = "black" 
        
    def add(self, a, b): 
        c = a + b 
        print(c) 

C1 = calci() 
print(C1.brand) 
print(C1.color) 
x = 10 
y = 20 
C1.add(x, y)
```

**With parameter with return value**

```python
class calci: 
    def __init__(self): 
        self.brand = "Casio" 
        self.color = "black"
    def add(self,a,b): 
        c=a+b 
        return c
C1 = calci()  
print(C1.brand) 
print(C1.color) 
x = 10 
y = 20 
res = C1.add(x,y) 
print(res)
```
