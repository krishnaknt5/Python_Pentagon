```python
class Farmer:
    def __init__(self, P, t, r):
        self.Principal = P
        self.time = t
        self.rate = r
        self.interest = (P*t*r)/100

f1 = Farmer(15000, 1, 2.5)
f2 = Farmer(35000, 5, 2.5)
print("Principal:", f1.Principal)
print("Interest:", f1.interest)
print("Principal:", f2.Principal)
print("Interest:", f2.interest)
```
