**@static method**

```python
class Car: 
    def __init__(self): 
        self.color = "black" 
        
    def Protect(self): 
        print("Car protect us from rain and sun.")
        #staticmethod
    @staticmethod
    def travel():
        print("WE r travelling")
        #CLASS METHOD
    @classmethod
    def milage(cls):
        print("MIlage is 20-25")   

C1 = Car() 
print(C1.color)
C1.Protect()
C1.travel()
C1.milage()
Car.travel()
Car.milage()
# Car.Protect()
```
