**Functions**

**no parameter no return value**

```python
def add():
    a=10
    b=20
    c=a+b
    print(c)
add()
```

**no parameter no return value**

```python
def add():
    a=10
    b=20
    c=a+b
    return c
result=add()
print(result)
```

**with parameter no return value**

```python
def add():
    a=10
    b=20
    c=a+b
    return c
result=add()
print (result)
```

**with parameter with return value**

```python
def add(a,b):
    c=a+b
    return c
result=add(a=1,b=2)
print(result)
```
